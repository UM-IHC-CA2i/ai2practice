### Values and variables

A **variable** is a name that stores a value, like a labeled specimen jar. You create one with `=`, which means "store the value on the right under the name on the left" (not "is equal to"). Values come in **types**: whole numbers (`int`), decimal numbers (`float`), text (`str`, short for string, always in quotes) and `True`/`False` (`bool`). `print()` shows a value under the cell. Lines starting with `#` are **comments**: notes for humans that Python ignores.

````python
modality = "CT"          # text (a string)
studies_today = 38       # a whole number
mean_dose_mGy = 12.5     # a decimal number
contrast_given = True    # True or False

print(modality, studies_today, mean_dose_mGy, contrast_given)
print(type(studies_today), type(mean_dose_mGy))
print("Studies tomorrow if we do 5 more:", studies_today + 5)
````

### Functions

A **function** is a named, reusable piece of code. You **define** it once with `def`, give it inputs in brackets (called **arguments**), and it hands back a result with `return`. Then you can **call** it as often as you like. You have already used built-in functions such as `print()` and `len()`.

````python
def mean_of(values):
    """Return the average of a list of numbers."""
    return sum(values) / len(values)

print(mean_of([38, 41, 35, 44, 40]))
````

### Counting and filtering

`value_counts()` counts how many rows have each value in a column, most common first. To **filter** (keep only some rows), put a condition in square brackets: `df[df["ordering_dept"] == "ED"]` keeps only the rows where the department is ED. Note the double `==`, which asks "is it equal?", unlike the single `=` that stores a value.

````python
print(df["ordering_dept"].value_counts())

ed = df[df["ordering_dept"] == "ED"]
print("Studies ordered by the ED:", len(ed))
````

## Part 3: A first chart

A chart often shows a pattern faster than a table. We use **matplotlib**, the standard Python plotting library (imported as `plt`). The pattern is always the same: make a figure and its **axes** (the plotting area, stored as `ax`), draw on `ax`, add a title and axis labels, then `plt.show()`.

````python
dept_counts = df["ordering_dept"].value_counts()

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(dept_counts.index, dept_counts.values, color="#0D9488")
ax.set_title("CT studies by ordering department, October 2026")
ax.set_xlabel("Ordering department")
ax.set_ylabel("Number of studies")
plt.show()
````
